## 딕셔너리 (Dictionary)

- {"키":"값"} # 키는 내부 해시테이블로 관리하고 찾음(인덱스값과 비슷한 느낌)
- "키" 중복 불가, 단, 값은 중복 가능
- 데이터 추가하는 순서를 유지(파이썬 특징)


```python
{"a":1}
{"a":1,"b":1} # 키 중복 불가, 값 중복 가능
print({"a":1, "b":2})
print({"a":1, "b":1})
```

```python
user_info={
    "name" : "LeeJiHye",
    "age" : 29
}

print(user_info)

print(user_info["age"]) # 키로 값을 가져오기

user_info["age"] = 30

print(user_info["age"]) # 키에 따른 값 변경 가능

user_info["role"] = "student" # 인덱서 방식 : 없는 키에 값을 할당하여 추가 가능

print(user_info)

# user_info[0] 처럼 인덱스 번호를 넣는 식으로는 값을 가져올 수 없다
```

### 삭제

```python
user_info={
    "name" : "LeeJiHye",
    "age" : 29,
    "role" : "admin" # 변수가 전역적으로 있을 때는 반드시 실행해야 반영된다
}

print(user_info)

# 1. del 메서드 사용 방식
del user_info["role"] # del 메서드로 삭제, 혹시 값으로는??삭제 가능?

print(user_info)

# 2. .pop(키값) 사용 방식(값을 반환하며 삭제)
age = user_info.pop("age")

print(age, user_info)
```

### 딕셔너리 주요 메서드
- keys(), values(), items()

```python
menu = {"coffee":4000, "tea":4500}

# key 목록(리스트) 조회
print(menu.keys()) # 출력 결과물 : dict_keys(['coffe', 'tea']) <- 리스트

# value 목록(리스트) 조회
print(menu.values())

# items 목록(키값쌍 튜플들의 리스트) 조희
print(menu.items())

# 딕셔너리는 인덱서 방식으로 값 참조 불가 -> 리스트로 변환하여 참조하는 것은 가능
print(list(menu.keys())[1]) # list : ()안의 parameter를 list로 변환

# sum 계산함수로 딕셔너리 내 값들의 합을 출력 가능
print(sum(menu.values()))
```

```python
# 딕셔너리 내 키가 있는지 여부 확인(in은 key만 검색 가능!)
print("coffee" in menu)

# 값을 검색하려면 딕셔너리.values()를 대상으로 확인
print(4000 in menu.values())
```

```python
# 없는 key를 찾으면 KeyError 발생
# menu["cake"] 

# get으로 keyerror 없이 default 값 설정 가능
price = menu.get("cake",5000)
print(price)

# get이 기존 딕셔너리에 새로 데이터를 추가를 해주진 않음!
print(menu)

# default 값이 없는 경우 None으로 출력
print(menu.get("cake"))

# 기존에 원래 key가 있던 경우
print(menu.get("coffee",0)) # 기존에 있는 value값을 반환함. 기본값 0은 없을때만!(덮어쓰기x)
```

### 집합(Set)
- 수학의 집합과 동일한 개념. 중복과 순서가 없어서 인덱스로 꺼낼 수 없는 자료형
- {} 중괄호 사용
- set() 함수를 생성

```python
grade = set([1,2,2,3,3,3]) # 데이터 중복해서 넣어도

print(grade) # 결국 {1,2,3}만 출력됨
```

```python
print(type({})) # {} : 빈 딕셔너리

print(type(set())) # set() : 빈 집합

print(type({1,2,2})) # {1,2} 집합
```

```python
s = {1,2}

# .add()로 집합에 새 값 추가
s.add(2) # 기존에 있는 값 -> 2가 2개로 늘어나진 않음
s.add(5) # 새 값

print(s)

# .remove()로 값 삭제, 단, 없는 값을 삭제하려하면 KeyError 발생!
s.remove(1)
print(s)

# .discard()로도 값 삭제 가능, 없는 값 삭제해도 KeyError 발생하지 않음
s.discard(2)
print(s)
```

```python
# 집합 연산
set_a={1,2,3}
set_b={3,4,5}

# 교집합(A & B)
print(set_a & set_b)

# 합집합(A | B)
print(set_a | set_b)

# 차집합(A - B)
print(set_a - set_b)

# 포함 여부 확인
print(3 in set_a)
```

#### 중복제거 및 정렬

```python
# 중복된 데이터가 있는 기존 리스트
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]

# 1. set() : 집합으로 변환하여 중복 제거
unique_set = set(raw_emails) 

# 집합을 다시 리스트 형태로 변환
print(list(unique_set)) 

# 정렬
print(sorted(list(unique_set))) 

# [참고] 정렬 sorted
print(sorted(set(["c","d","a","f","a"])))

# 2. dict.fromkeys() : 중복제거 및 기존 데이터 순서를 유지
sort_emails = list(dict.fromkeys(raw_emails))
print(sort_emails)
```
