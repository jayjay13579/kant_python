### 논리 연산자
- and
- or
- not


````python
is_adult = True
has_ticket = False

# and 연산
print(is_adult and has_ticket) # false

# or 연산
print(is_adult or has_ticket) # true

# not 연산
print(not is_adult) # false
````

````python
print("기본 출력: %s, %d" % ("손흥민", 20)) # s: 문자열, d: 정수
print("기본 출력: [%5d]" % 20) # 왼쪽 5칸부터 정렬
print("기본 출력: [%-5d]" % 20) # 오른쪽 5칸부터 정렬
print("소수점 표기: %.2f" % 3.141592) # 소수점 2자리까지 표기, f: 실수
````

````python
name = "cake"
price = 6500
qty = 2

# format 방식
format_str = "주문내역 : {0} {1}  -> 총 {2}원".format(name, qty, price*qty)
print(format_str)
format_str = "주문내역 : {} {}  -> 총 {:,.2f}원".format(name, qty, price*qty)
print(format_str)

# f-string 방식
format_str = f"주문내역 : {name} {qty} -> 총 {(price*qty):<10,}원" #{}안 계산가능
print(format_str)

print("{:>30,.2f}".format(123456.4568))
````

````python
print(f"금액 : {15000:>8,}원")
# f : formating, > : 오른쪽 정렬(<:왼쪽), 8 : 8칸 확보, , : 천 단위 구분
````

````python
# 가상의 주문 데이터
orders = [ 
    ("Latte", 5000, 2),
    ("Americano", 3000, 1),
    ("Cake", 6000, 1)
]

print("="*36)

# 항목명 작성
print(f"{'메뉴명':<10} {'단가':>8} {'금액':>9}") # 문자열도 {}안에서 정렬 가능, 단, ' ' 처리

# 구분선
print("-"*36)


# 총 금액 계산
total_amount = 0
for menu, price, qty in orders: # orders 리스트 한 행 각 항목을 menu, price, qty로 받아옴
    subtotal = price * qty
    total_amount += subtotal
    print(f"{menu:<14} {price:>8} {subtotal:>10,}")

# 구분선
print("-"*36)

# 합계 금액
print(f"{'합계 금액:':<20} {total_amount:>9,}원")
````
