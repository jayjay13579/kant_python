### 변수 이름 규칙 
- 숫자로 시작 불가 : 1_name(x)
- 공백 불가 : user name (x)
- 하이픈(-) 불가 : user-age (x)
- 예약어 사용 불가 : if, def 등 사용 불가

### 컨벤션
- 스네이크 표기법 : user_name, value_list

### 다른 표기법
- 파스칼 케이스(키워드 첫 대문자) : UserName, ValueList
- 카멜 케이스 : userName

### garbage collector
- 변수가 할당되지 않은 객체, 미사용 객체는 모아서 나중에 메모리 release 하는 역할
- 대부분의 managed language - 메모리 자동 관리 언어들은 모두 있음
- 메모리 할당 해제를 개발자가 신경쓰지 않아도 됨


````python
x = 10
print(type(x))
print(id(x)) # 객체 '10'의 주소값
x = "hello"
print(type(x))
print(id(x))
````
