````python
# 리스트 : 가변 데이터 저장
shopping_list = ["apple", "banana", "cherry"]
print(shopping_list)
print(shopping_list[0])

shopping_list[0]="kiwi" # 인덱스 0의 값 수정(덮어쓰기)
print(shopping_list[0])
print(shopping_list)
````

#### 리스트 메서드


````python
fruits = ["apple"]

# append: 마지막에 추가
fruits.append("banana") 
print(fruits) # 메서드는 원본을 바꾼다

# insert: 원하는 인덱스 위치에 추가
fruits.insert(0,"kiwi") # insert(인덱스 넘버, 값)
print(fruits) 

# remove: 해당 값 삭제
fruits.remove("apple")
print(fruits) 

# fruits.remove("grape") : value error 발생
# 참고 : ctrl + / 로 멀티 라인 주석 처리 가능
````

### 튜플(tuple)
- 한번 생성하면 요소 추가, 수정, 삭제 불가


````python
axis = (38.123, 41.123)

# axis[0] = 36.123 # TypeError 발생

point = 1,2,3  # tuple은 () 생략 가능
print(point, type(point))
````

````python
import sys
print(sys.getsizeof([1,2,3])) # = 88
print(sys.getsizeof((1,2,3))) # = 72
# list 보다 tuple의 사이즈가 조금 더 작음
# 데이터 안정성이 중요할 때 tuple 사용
# 데이터 가변성이 중요할 때 list 사용
````

### 인덱스
```py
colors = ['red', 'green', 'blue']
```
- 정방향 : 0, 1, 2
- 역방향 : -3, -2, -1

````python
colors = ['red', 'green', 'blue']
print(colors[0])
print(colors[-1]) # len(list)-1 인덱스 계산없이 마지막 값 가져올 수 있음 
print(colors[1]==colors[-2])
# print(colors[3]) #IndexError 발생
````

### 슬라이싱

````python
numbers = [0,1,2,3,4,5]

# 1번부터 4번 미만까지
numbers[1:4]

# 2칸 간격으로 추출
numbers[::2]

# 역순으로 추출
numbers[::-1]

# 역순으로 2칸 간격 추출같은것도 되나?

# 앞에서 3개만 추출
numbers[:3]

# 인덱스 3번부터 끝까지
numbers[3:]
````
