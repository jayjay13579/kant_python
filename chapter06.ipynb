## 종합 실습

```python
# 회원 데이터
user_database = [
    {"name":"손흥민", "level":"VIP", "active":True, "point":15000},
    {"name":"김민재", "level":"Gold", "active":False, "point":9000},
    {"name":"조규성", "level":"VIP", "active":True, "point":12000},
    {"name":"이강인", "level":"Gold", "active":True, "point":8000},
    {"name":"홍명보", "level":"Bronze", "active":False, "point":1000}
]

# VIP 포인트 누적
total_vip_points = 0

# 현재 활성 회원 수
active_user_count = 0



print("=== 마케팅 대상 활성 VIP 및 Gold 회원 식별 결과 ===")

for user in user_database: # user는 딕셔너리 한 행
    # 활성화 된 고객 필터링 active
    if user["active"]:
        # 활성 회원 수 누적
        active_user_count += 1
        if user["level"] == "VIP":
            # vip 포인트 누적
            total_vip_points += user["point"]
            print(
                f"[{user['level']}등급] {user['name']}"
                f" 회원님에게 {user['level']}쿠폰이 발행되었습니다."
                ) # ""안에서 'name'으로 작성
        elif user["level"] == "Gold":
            print(
                f"[{user['level']}등급] {user['name']}"
                f" 회원님에게 {user['level']}쿠폰이 발행되었습니다."
                ) # ctrl + D 멀티커서로 여러개 한번에 수정


print(f"전체 활성 회원 수 : {active_user_count}명")
print(f"활성 VIP 포인트 누적 : {total_vip_points}점")
```

```python
# 숫자 맞추기 게임: 맞출때까지 while 문 돌면서 업, 다운 제시, 맞추면 break로 탈출

import random  # 난수 발생을 위한 random 모듈 import

# 1~100 사이의 난수 발생
number = random.randint(1, 100)

# 시도 횟수 count
try_count = 0

# 타이틀
print("="*40)
print("컴퓨터가 1부터 100사이의 숫자를 생성했습니다.")
print("="*40)

# 메인 루프
while True:
    user_input = input("컴퓨터의 숫자를 맞춰보세요(1~100): ") # 여기에 바로 int 씌우면 문자열 넣었을 때 오류 발생

    # q 누르면 게임 종료
    if user_input == "q":
        print(f"게임 종료합니다. 정답은 {number}입니다.")
        break
    
    # 입력값 숫자인지 확인(isdecimal 메서드 : 숫자는 true 확인)
    if not user_input.isdecimal():  #==False 라고 해도 됨?
        print("숫자만 입력해주세요.")
        continue # 다시 while 루프로 돌아가서 값 계속 입력받도록

    user_number = int(user_input)

    # 1~100 사이가 아닌 값을 입력하면
    if user_number >100 or user_number <1 : # 문자열과 숫자는 비교할 수 없음. 이 코드 위치가 int(user_input)아래 오도록
        print("1~100사이의 값을 입력해주세요.")
        continue

    # 시도 횟수 증가
    try_count += 1

    # 입력값과 정답을 비교
    if user_number < number:
        print("Up! 더 큰 수입니다.")
    elif user_number > number:
        print("Down! 더 작은 수입니다.")
    else:
        print(f"{number} 정답입니다! {try_count}회만에 성공했습니다.")
        break # 정답일 경우 무한루프 탈출
```

```python
# 난수 생성을 위한 random 모듈 import
import random

# 1~100 사이의 난수 생성하고 number 변수 할당
number = random.randint(1,100) # 100까지도 포함! cf) .randrange(1,100)은 100 제외

# 시도 횟수 count
try_count = 0

# 타이틀
print("="*40)
print("컴퓨터가 1부터 100 사이의 숫자를 생성했습니다.")
print("="*40)

# 메인 루프 while문 작성
while True:
    # 주의 : 바로 int(input()) 씌워버리면 아래에 문자열 관련 조건에서 Error 발생 🚨
    user_input = input("컴퓨터의 숫자를 맞춰보세요(1~100): ")

    # 문자열 q 입력시 종료
    # 주의 : 이 조건이 입력값 숫자 검증 조건 뒤에 오면 안됨! 숫자 입력하라고 떠버림! 🚨
    if user_input == "q" :
        print(f"게임을 종료합니다. 정답은{number}입니다.")
        break # 게임 종료

    # 숫자 검증 : 문자열.isdecimal()은 빈 값이 아니고 모든 문자가 십진숫자 문자면 True
    if not user_input.isdecimal():
        print("숫자만 입력해주세요.")
        continue # 다시 루프돌며 값 입력받도록! 게임 계속 진행

    # 숫자 검증 되었을 때 int로 변환
    user_number = int(user_input) 

    # 범위 검증 : 1~100 사이가 아닌 값을 입력하면 안내 메시지
    if user_number > 100 or user_number < 1 :
        print("1~100 사이의 값을 입력해주세요")
        continue # 다시 루프돌며 값 입력받도록! 게임 계속 진행
    
    # 시도 횟수 카운트
    # 주의 : 값 비교 부분 뒤에 적으면 (break 때문에) 정답 맞힌 시도 1회가 카운트 안됨 🚨
    try_count += 1

    # 숫자 비교
    if user_number < number:
        print("Up! 더 큰 수입니다.")
    elif user_number > number:
        print("Down! 더 작은 수입니다. ")
    else: # 어차피 else에는 같은 경우만 걸러짐
        print(f"{number} 정답입니다! {try_count}회만에 성공했습니다.")
        break # 정답 맞히면 게임 종료! 반드시 break로 무한루프 탈출해줘야 함
```
