## 함수(function)
- 코드의 재사용성을 높이기 위해 반복되는 로직을 합쳐놓은 문법
- 매개변수(parameter)를 받아 작동
- 단독사용 가능(ex: print(), len() 내장함수 등)

```py
def 함수이름(매개변수):
    실행 로직1
    실행 로직2
    실행 로직3
```

````python
# 함수 정의
def say_hello(name, age, point=1000):
    # point default값 지정, 단, default 변수는 무조건 마지막에 와야 함🚨
    print(f"안녕하세요. 저는 {name}입니다. 나이는 {age}입니다.")
    print(f"초기 지급 포인트 {point}점이 지급되었습니다.")

# 함수 호출
say_hello("LeeJiHye", 29) # 미입력 변수는 (있으면) default값 반환
say_hello("LeeJiHye", 29, 2000) # 입력되면 default 무시하고 입력값 반환
````

````python
# 가변 인수 : 인수 개수 알 수 없을 때 -> *를 붙임
# parameter 관례(*args)

def add_all(*numbers): # numbers 매개변수(parameter)의 개수는 가변적(*)
    print(f"입력된 튜플 : {numbers}")
    total = sum(numbers)
    print(f"합계 결과 : {total}")

add_all(1,2,3) # 인수(argument) 넣을 때 괄호 생략해도 튜플

add_all(1,2,3,4,5)

add_all() # 인수 넣지 않아도 오류 X.. 가변인수 특징?
````

````python
# 딕셔너리 가변 인수 : **로 표현(관례 **kwargs)

def user_info(**data):
    print(f"전달된 딕셔너리 구조 : {data}")
    for key, value in data.items():
        print(f"{key}: {value}")

# 함수 호출
user_info(name="스파르타", level=5, active=True) # 왜 : 안씀?
````

````python
# return - 결과값 반환
# 하나의 반환값

def check_ord_number(number):
    if number > 0:
        return "양수"
    elif number == 0: # 비교할때 == 두번쓰는거 제발 잊지마! 🚨
        return "0"
    else:
        return "음수"

print(check_ord_number(10))
print(check_ord_number(-20))
print(check_ord_number(0))
````

````python
# 여러개 결과값 반환
def calc_value(a,b):
    plus = a+b
    minus = a-b
    return plus, minus # 콤마 seperator, 자동으로 튜플로 리턴

result1, result2 = calc_value(10,5) # 튜플을 다시 2개의 변수로 나눠 값 배정
print(f"덧셈 : {result1}, 뺄셈 : {result2}")

result = calc_value(10,5)
print(f"튜플 : {result}")

type(result)
````

````python
# 평균 계산해서 반환하는 함수

def get_average(*numbers):
    if len(numbers) == 0 :
        return 0
    # 혹은 if not scores :
    #         return  
    total = sum(numbers)
    numbers_average = total / len(numbers)
    return numbers_average

result1 = get_average(80, 90, 100)
result2 = get_average(75, 80, 92, 100, 55)
result3 = get_average() # ZeroDivision 오류 발생 예방을 위해 함수 내 if조건 추가

print(result1, result2, result3)
````

````python
# 미니 계산기
# 사칙 연산기

def add(a,b):
    return a+b

def minus(a,b):
    return a-b

def multiply(a,b): 
    return a*b

def divide(a,b):
    if b==0:
        return "계산 불가, 0으로 나눌 수 없습니다."
    return a/b

a=input("첫번째 숫자:")
b=input("첫번째 숫자:")

print(f"덧셈 연산 : {a} + {b} = {add(int(a),int(b))}")
print(f"뺄셈 연산 : {a} - {b} = {minus(int(a),int(b))}")
print(f"곱셈 연산 : {a} * {b} = {multiply(int(a),int(b))}")
print(f"나눗셈 연산 : {a} / {b} = {divide(int(a),int(b))}")
````

````python
# 전역변수
value = "전역 데이터"

# fuction ================
def outer_function():
    value = "지역 변수 1"

    def  inner_function():
        value = "지역 변수 2"
        print(value)

    inner_function()
    print(value)
# fuction ================

outer_function()
print(value)
````

````python
# global 키워드
counter = 0

def inc_counter():
    global counter # global로 전역변수 참조 선언
    counter += 1

inc_counter() # counter = 1
inc_counter() # counter = 2
print(counter)
````

````python
# global 키워드
counter = 0

def inc_counter():
    counter = 0
    # global counter # global로 전역변수 참조 선언
    counter += 1

inc_counter() # 지역 counter = 1
inc_counter() # 지역 counter = 2
print(counter) # 전역 counter = 0 
````

## Lambda Expression 람다식
- 일종의 함수, 1줄로 간결하게 함수 작성하는 방식임
- 과도하게 사용하면 가독성이 떨어짐
- return 키워드 없이도 결과값 자동 반환

```py
lambda 매개변수: 표현식(로직)
```

````python
# 일반함수 

def add_five(x):
    return x + 5

print(add_five(10))

# lambda식

result = lambda x : x + 5  # x 가 매개변수
 
print(result(10)) # result는 lambda 식을 담고 있는 일종의 함수, 뒤에 ()인수 넣어줘야 함
````

````python
# map 모든 요소를 반환

numbers = [1,2,3,4,5]

result = list(map(lambda x : x+10, numbers))
# numbers의 모든 인자들을 map으로 lambda식에 모두 넣은 후 순서대로 결과값 반환
# for문 없이 반복처리 가능

print(f"계산결과 : {result}")

# 리스트 컴프리핸션 (???)
print([x+10 for x in numbers]) # for문을 한 줄로 쓰고 결과를 list로 받음, 동일한 결과

filter_result = list(filter(lambda x : x % 2 == 0, numbers)) #만약 list 안적으면?
# filter로 if 없이 해당 조건에 맞는(True인) x만 추출

print(f"짝수 결과 : {filter_result}")

print([x for x in numbers if x % 2 ==0]) # for문, if문을 한 줄로 쓰고 결과를 list로 받음, 동일한 결과
````

````python
# lambda식으로 정렬

players = [
    ("손흥민", 95),
    ("김민재", 60),
    ("이강인", 80)
]
print(sorted(players, key=lambda p : p[1])) # 내림차순 # 만약 이걸 lambda를 안쓰면 어떻게 씀?
print(sorted(players, key=lambda p : p[1], reverse=True)) # 점수로 sorting(reverse=True 내림차순)
print(max(players, key=lambda p : p[1])) # 최고 점수 추출 
````

````python
numbers = [2,3,4,5,6,7]

# lambda 식으로 모든 요소의 거듭제곱 추출
result = list(map(lambda x : x**2, numbers))
print(f"거듭제곱 결과값 : {result}")

# 결과값이 20 이상인 값만 추출
result2 = list(filter(lambda x : x >= 20, result))
print(f"거듭제곱값이 20 이상 : {result2}")
````
